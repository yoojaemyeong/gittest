## 생성형 AI(LLM, 대규모 언어 모델) 특징1
* 알고리즘적 추론과 인간 사고 사이의 중간 영역에 위치한 모델
* LLM은 자연어에 유창하나 내재적 의미를 이해하고 수행하지는 않는다.
  > 알고리즘의 특징은 절차적이고 모호성이 없다.    
  > 인간 사고는 창의적이고 모호성을 포함한다.  
* LLM은 이 두 영역 사이에서 자연어 기반 추론을 수행하는 중간 지능

## 생성형 AI(LLM, 대규모 언어 모델) 특징2
* 방대한 문서와 데이터에 분산되어 있는 지식을 트랜스포머 모델로 학습 > 구조화·요약·재조합 > 사람이 필요로 하는 형태로 지식을 제공하는 기술
* 페르소나(Persona): 조건부 확률로 응답 성향 편향(문체, 역할, 전문성 범위 등을 유도)
* 할루시네이션: 사실과 무관한 내용을 그럴듯하게 생셩하는 현상
* 확률적 생성 모델(Probabilistic  Generator)
* 사실 인식 능력 부재: 다시 말해 정확성은 학습데이터와 컨텍스트에 의존적, 외부 검증없이는 신뢰할 수 없음
* LLM의 출력은 입력 컨텍스트에 강하게 의존적( 질문 순서, 표현 방식, 이전 대화가 결과에 영향을 줌)
* LLM의 내부 추론 과정은 명시적 해석 불가(사후 생성된 언어적 설명)
* 제약조건: Temperature, Top-p, 토큰 길이 등 파라미터에 민감
* RAG(Retrieval-Augmented Gengeration), MoE(Mixture of Expert)는 할루시네이션 완화의 핵심 구조
* LLM은 지식 엔진이 아니라 언어 생성 엔진이다. 곧 **정답 제공자가 아니라 추론 보조 도구에 가깝다.**

## LLM 설정 항목
* **Temperature** : 창의성 파라미터(안정적 0 ~ 2.0 창의성)
* **Top P**: 확률적 샘플링 0 ~ 1 (0.3:공식 문서 작성, 0.9:자유로운 설명, 스토리 작성)
* **Top k**: 확률 높은 단어 상위 몇 개만 사용하도록 제한(10:정형화된 설명, 50:다양한 표현 허용)
* **Maximum length**(토큰 수): 100(한 문단 요약), 800(강의용 상세 설명)
* **Frequency / Repeat Penalty**: 같은  단어 또는 문장을 계속 반복하지 않도록 조절(기본값: 일반 대화, 반복억제강화: 보고서·장문 작성)
* **Stop Sequence**(정지 토큰): 특정 단어가 나오면 AI가 바로 말을 멈추게 함(Stop:"###" -> 제목 단위로 출력 종료, Stop:"END" -> 응답 자동 종료)
